# Bước 6: Phân cụm khách hàng từ Luật kết hợp (Association Rules → Clustering)

Notebook này lấy **kết quả luật kết hợp** (Apriori/FP-Growth) và biến chúng thành **đặc trưng** để phân cụm khách hàng bằng K-Means.

## Ý tưởng cốt lõi
- Mỗi luật có dạng: **Antecedent → Consequent**
- Với mỗi khách hàng, ta kiểm tra: khách đó đã từng mua **đủ antecedents** của luật hay chưa.
- Mỗi luật trở thành một feature (0/1 hoặc có trọng số theo lift/confidence).
- (Tuỳ chọn) Ghép thêm **RFM** để phân cụm ổn định hơn.


## Parameters
Gán tham số để chạy bằng papermill.


```python
# PARAMETERS (for papermill)

# Input
CLEANED_DATA_PATH = "data/processed/cleaned_uk_data.csv"
RULES_INPUT_PATH = "data/processed/rules_apriori_filtered.csv"  # hoặc rules_fpgrowth_filtered.csv

# Feature engineering
TOP_K_RULES = 200
SORT_RULES_BY = "lift"      # lift | confidence | support
WEIGHTING = "lift"          # none | lift | confidence | support | lift_x_conf
MIN_ANTECEDENT_LEN = 1
USE_RFM = True
RFM_SCALE = True
RULE_SCALE = False

# Clustering
K_MIN = 2
K_MAX = 10
N_CLUSTERS = None            # None => chọn theo silhouette, hoặc đặt số cụ thể (vd 5)
RANDOM_STATE = 42

# Output
OUTPUT_CLUSTER_PATH = "data/processed/customer_clusters_from_rules.csv"

# Visual
PROJECTION_METHOD = "pca"   # pca | svd
PLOT_2D = True
```

## Set up


```python
%load_ext autoreload
%autoreload 2

import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Determine correct project root
cwd = os.getcwd()
if os.path.basename(cwd) == "notebooks":
    project_root = os.path.abspath("..")
else:
    project_root = cwd

src_path = os.path.join(project_root, "src")
if src_path not in sys.path:
    sys.path.append(src_path)

from cluster_library import RuleBasedCustomerClusterer
```

# Load cleaned data & rules


```python
CLEANED_DATA_PATH = "../data/processed/cleaned_uk_data.csv" 

df_clean = pd.read_csv(CLEANED_DATA_PATH, parse_dates=["InvoiceDate"])
print(df_clean.shape)
df_clean.head()
```

```python
RULES_INPUT_PATH = "../data/processed/rules_apriori_filtered.csv"
clusterer = RuleBasedCustomerClusterer(df_clean=df_clean)
customer_item_bool = clusterer.build_customer_item_matrix(threshold=1)
print('Customer × Item:', customer_item_bool.shape)

rules_df = clusterer.load_rules(
    rules_csv_path=RULES_INPUT_PATH,
    top_k=TOP_K_RULES,
    sort_by=SORT_RULES_BY,
)
print('Rules used:', rules_df.shape)
rules_df.head()
```

## Build features (Rules → Features) + (optional) RFM


```python
X, meta = clusterer.build_final_features(
    weighting=WEIGHTING,
    use_rfm=USE_RFM,
    rfm_scale=RFM_SCALE,
    rule_scale=RULE_SCALE,
    min_antecedent_len=MIN_ANTECEDENT_LEN,
)
print('X shape:', X.shape)
meta.head()
```

## Choose K (silhouette)


```python
sil_df = clusterer.choose_k_by_silhouette(
    X,
    k_min=K_MIN,
    k_max=K_MAX,
    random_state=RANDOM_STATE,
)
sil_df
```

```python
best_k = int(sil_df.loc[0, 'k'])
k = best_k if N_CLUSTERS is None else int(N_CLUSTERS)
print('Chosen k =', k)
```

## Fit KMeans & save results


```python
labels = clusterer.fit_kmeans(X, n_clusters=k, random_state=RANDOM_STATE)
meta_out = meta.copy()
meta_out['cluster'] = labels

# Lưu
os.makedirs(os.path.dirname(OUTPUT_CLUSTER_PATH), exist_ok=True)
meta_out.to_csv(OUTPUT_CLUSTER_PATH, index=False)
print('Saved:', OUTPUT_CLUSTER_PATH)
meta_out.head()
```

## Quick profiling


```python
profile_cols = ['cluster'] + ([c for c in ['Recency','Frequency','Monetary'] if c in meta_out.columns])
summary = meta_out.groupby('cluster').agg({
    'CustomerID': 'count',
    **{c:'mean' for c in profile_cols if c!='cluster'}
}).rename(columns={'CustomerID':'n_customers'}).sort_values('n_customers', ascending=False)
summary
```

## 2D visualization (PCA/SVD)


```python
if PLOT_2D:
    Z = clusterer.project_2d(X, method=PROJECTION_METHOD, random_state=RANDOM_STATE)
    plt.figure(figsize=(8,6))
    plt.scatter(Z[:,0], Z[:,1], c=labels, s=10)
    plt.title('Customer clusters (2D projection)')
    plt.xlabel('Component 1')
    plt.ylabel('Component 2')
    plt.tight_layout()
    plt.show()
```

```python
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# --- CẤU HÌNH 1: BASELINE (Binary Rules, No RFM) ---
print("--- ĐANG CHẠY BIẾN THỂ 1: BASELINE ---")

# 1. Tạo đặc trưng
X_v1, meta_v1 = clusterer.build_final_features(
    weighting=False,
    use_rfm=False,
    min_antecedent_len=1
)

# 2. Tìm K tối ưu
best_score_v1 = -1
best_k_v1 = 0
scaler = StandardScaler()
X_v1_scaled = scaler.fit_transform(X_v1) # Chuẩn hóa dữ liệu ngay tại đây

print(f"Kích thước dữ liệu V1: {X_v1.shape}")
print("Kết quả Silhouette Score cho V1:")

for k in [3, 4, 5, 6]:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_v1_scaled)
    
    # Tính điểm
    score = silhouette_score(X_v1_scaled, labels)
    print(f"  - K={k}: {score:.4f}")
    
    if score > best_score_v1:
        best_score_v1 = score
        best_k_v1 = k

print(f"=> Cấu hình 1 tốt nhất tại K={best_k_v1} với điểm = {best_score_v1:.4f}\n")
```

```python
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# --- CẤU HÌNH 2: ADVANCED (Weighted Rules + RFM) ---
print("--- ĐANG CHẠY BIẾN THỂ 2: ADVANCED ---")

# 1. SỬA LỖI: Tính TotalAmount trực tiếp trên df_clean bên ngoài
if 'TotalAmount' not in df_clean.columns:
    df_clean['TotalAmount'] = df_clean['Quantity'] * df_clean['UnitPrice']
    print("Đã tạo cột TotalAmount trên df_clean.")

# Lưu ý: Nếu class RuleBasedCustomerClusterer của bạn không tự cập nhật dữ liệu mới
# Bạn có thể cần gán lại (nhưng cứ thử chạy trước xem có lỗi không)
# clusterer.data = df_clean (nếu thuộc tính tên là data)
# clusterer.df_clean = df_clean (nếu thuộc tính tên là df_clean)

# 2. Tạo đặc trưng (Weighted + RFM)
X_v2, meta_v2 = clusterer.build_final_features(
    weighting=True,
    use_rfm=True,
    rfm_scale=True,
    rule_scale=True,
    min_antecedent_len=1
)

# 3. Tìm K tối ưu
best_score_v2 = -1
best_k_v2 = 0
scaler = StandardScaler()
X_v2_scaled = scaler.fit_transform(X_v2) 

print(f"Kích thước dữ liệu V2: {X_v2.shape}")
print("Kết quả Silhouette Score cho V2:")

for k in [3, 4, 5, 6]:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_v2_scaled)
    
    score = silhouette_score(X_v2_scaled, labels)
    print(f"  - K={k}: {score:.4f}")
    
    if score > best_score_v2:
        best_score_v2 = score
        best_k_v2 = k

print(f"=> Cấu hình 2 tốt nhất tại K={best_k_v2} với điểm = {best_score_v2:.4f}\n")
```

```python
print("=== TỔNG KẾT SO SÁNH ===")
print(f"Biến thể 1 (Cơ bản): Điểm cao nhất = {best_score_v1:.4f} (tại K={best_k_v1})")
print(f"Biến thể 2 (Nâng cao): Điểm cao nhất = {best_score_v2:.4f} (tại K={best_k_v2})")

if best_score_v2 > best_score_v1:
    print("=> KẾT LUẬN: Biến thể 2 (Nâng cao) cho kết quả tách cụm tốt hơn.")
    final_X = X_v2
    final_meta = meta_v2
    final_k = best_k_v2
else:
    print("=> KẾT LUẬN: Biến thể 1 (Cơ bản) cho kết quả tách cụm tốt hơn (hoặc tương đương).")
    final_X = X_v1
    final_meta = meta_v1
    final_k = best_k_v1

print(f"Hệ thống sẽ chọn cấu hình tốt hơn (K={final_k}) để chạy mô hình cuối cùng...")

# Chạy lại mô hình với cấu hình tốt nhất để lưu kết quả cuối cùng
final_labels = clusterer.fit_kmeans(final_X, n_clusters=final_k, random_state=42)
final_meta['cluster'] = final_labels

# Hiển thị biểu đồ 2D cho cấu hình thắng cuộc
import matplotlib.pyplot as plt
Z = clusterer.project_2d(final_X)
plt.figure(figsize=(10,7))
plt.scatter(Z[:,0], Z[:,1], c=final_labels, cmap='viridis', s=15)
plt.title(f'FINAL RESULT: Cluster Visualization (K={final_k})')
plt.colorbar(label='Cluster Label')
plt.show()
```

```python
import pandas as pd

print("=== KẾT QUẢ SO SÁNH ===")
print(f"Biến thể 1 (Cơ bản): {best_score_v1:.4f} (K={best_k_v1})")
print(f"Biến thể 2 (Nâng cao): {best_score_v2:.4f} (K={best_k_v2})")

if best_score_v2 > best_score_v1:
    print("\n=> CHỌN: Biến thể 2 (Nâng cao) vì có chỉ số tách cụm tốt hơn.")
    final_meta = meta_v2.copy()
    final_k = best_k_v2
    # Cần gán lại nhãn cluster cho final_meta theo K tốt nhất của V2
    # (Lưu ý: đoạn code chạy vòng lặp trước đó có thể đang lưu label của k cuối cùng là 6, nên cần chạy lại fit chuẩn xác)
    print(f"Đang chạy lại K-Means với K={final_k} cho cấu hình 2...")
    kmeans_final = KMeans(n_clusters=final_k, random_state=42, n_init=10)
    final_labels = kmeans_final.fit_predict(X_v2_scaled)
    final_meta['cluster'] = final_labels
else:
    print("\n=> CHỌN: Biến thể 1 (Cơ bản) vì tốt hơn hoặc tương đương.")
    final_meta = meta_v1.copy()
    final_k = best_k_v1
    print(f"Đang chạy lại K-Means với K={final_k} cho cấu hình 1...")
    kmeans_final = KMeans(n_clusters=final_k, random_state=42, n_init=10)
    final_labels = kmeans_final.fit_predict(X_v1_scaled)
    final_meta['cluster'] = final_labels

# 2. Lập bảng Profiling (Phân tích đặc điểm cụm)
print("\n=== BẢNG HỒ SƠ KHÁCH HÀNG (PROFILING) ===")

# Các cột cần tính trung bình
cols_to_agg = {'CustomerID': 'count'} # Đếm số lượng khách

# Thêm RFM vào tính toán (nếu có trong dữ liệu)
for col in ['Recency', 'Frequency', 'Monetary', 'TotalAmount']:
    if col in final_meta.columns:
        cols_to_agg[col] = 'mean'

# Thêm Top 3 luật đặc trưng (Tính trung bình giá trị của các cột Rule để xem cụm nào nổi bật luật nào)
# Lấy danh sách cột bắt đầu bằng 'Rule_'
rule_cols = [c for c in final_meta.columns if c.startswith('Rule_')]
for col in rule_cols:
    cols_to_agg[col] = 'mean'

# Gom nhóm và tính toán
summary = final_meta.groupby('cluster').agg(cols_to_agg).rename(columns={'CustomerID': 'n_customers'})
summary = summary.sort_values('n_customers', ascending=False)

# Chỉ hiển thị các cột RFM và Top 3 luật mạnh nhất cho mỗi cụm để dễ nhìn
display_cols = ['n_customers'] + [c for c in ['Recency', 'Frequency', 'TotalAmount'] if c in summary.columns]
print("Thống kê cơ bản:")
print(summary[display_cols].round(2))

print("\n--- ĐẶC ĐIỂM LUẬT MUA SẮM (TOP RULES) ---")
# Với mỗi cụm, tìm ra luật nào có giá trị trung bình cao nhất (tức là khách trong cụm đó hay thỏa mãn luật đó nhất)
for cluster_id in summary.index:
    print(f"\nCụm {cluster_id} (Số khách: {summary.loc[cluster_id, 'n_customers']}):")
    # Lấy giá trị trung bình các cột Rule của cụm này
    cluster_rules = summary.loc[cluster_id, rule_cols]
    # Top 3 luật cao nhất
    top_rules = cluster_rules.sort_values(ascending=False).head(3)
    
    for rule_name, score in top_rules.items():
        if score > 0.1: # Chỉ hiện nếu có chỉ số đáng kể
            print(f"  - {rule_name}: {score:.4f} (Độ phổ biến/Trọng số)")
        else:
            print("  - (Không có luật nào nổi bật)")
            break
```
